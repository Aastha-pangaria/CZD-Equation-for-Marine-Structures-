```python
import pandas as pd
import matplotlib.pyplot as plt

input_folder = "Final_Level_Data"

levels_to_check = [8, 9, 10]

data = {}

for level in levels_to_check:
    file_path = f"{input_folder}/Final_data_S1_L{level}.xlsx"
    df = pd.read_excel(file_path)
    data[level] = df

print("✅ Data loaded")
```

```python
for level, df in data.items():

    print(f"\n📊 Level {level}")

    plt.figure()
    plt.scatter(df["Twet"], df["IDMT"])
    plt.xlabel("Twet")
    plt.ylabel("IDMT")
    plt.title(f"Level {level}: IDMT vs Twet")
    plt.show()

    plt.figure()
    plt.scatter(df["Tdry"], df["IDMT"])
    plt.xlabel("Tdry")
    plt.ylabel("IDMT")
    plt.title(f"Level {level}: IDMT vs Tdry")
    plt.show()

    plt.figure()
    plt.scatter(df["J"], df["IDMT"])
    plt.xlabel("J")
    plt.ylabel("IDMT")
    plt.title(f"Level {level}: IDMT vs J")
    plt.show()

    plt.figure()
    plt.scatter(df["TEQ"], df["IDMT"])
    plt.xlabel("TEQ")
    plt.ylabel("IDMT")
    plt.title(f"Level {level}: IDMT vs TEQ")
    plt.show()
```

```python
for level, df in data.items():

    print(f"\n🔍 Correlation Matrix — Level {level}")

    corr = df[["Twet", "Tdry", "J", "TEQ", "IDMT"]].corr()

    print(corr)
```

```python
import seaborn as sns

for level, df in data.items():

    plt.figure(figsize=(6,4))
    sns.heatmap(df[["Twet", "Tdry", "J", "TEQ", "IDMT"]].corr(),
                annot=True, cmap="coolwarm")
    plt.title(f"Correlation Heatmap — Level {level}")
    plt.show()
```
